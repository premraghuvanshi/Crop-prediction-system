```python
X = df[['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']]
```

```python
y = df["label"]
```

```python
X_train, X_test ,y_train , y_test = train_test_split(X , y , test_size=0.3,random_state=42, stratify=y)
```

```python
numerical_feat = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']
preproccesor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(),numerical_feat)
        ]
)
```

```python
pipline =Pipeline(
    steps=[
        ("preproccesor", preproccesor),
        ("model", RandomForestClassifier(n_estimators=100, random_state=42, class_weight="balanced"))
    ]
)
```

```python
pipline.fit(X_train,y_train)
```

```python
y_pred = pipline.predict(X_test)
probabilities = pipline.predict_proba(X_test)
mean_test_confidence= probabilities.max(axis=1).mean()*100
test_accuracy=accuracy_score(y_test,y_pred)*100
report = classification_report(y_test,y_pred)

```

```python
print("***************Model Diagnostic********************")
print("confidence:",mean_test_confidence)
print("accuracy:", test_accuracy)
print("classification_report:", report)
```

```python
joblib.dump(pipline, 'Crop_recommendation.pkl')
```
