```python
import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder , RobustScaler
from sklearn.metrics import mean_absolute_error , root_mean_squared_error , r2_score , mean_squared_error
from xgboost import XGBRegressor
import joblib
```

```python
df = pd.read_csv("D:\Crop Prediction System\Model\Datasets\crop_production.csv")
df
```

```python
df1=df[df["State_Name"] == "Madhya Pradesh"].reset_index(drop=True)

df1
```

```python
df1.drop(columns=["State_Name"], inplace=True)
df1
```

```python
df1.isnull().sum()
```

```python
df1.isnull().mean()*100
```

```python
df1.dropna(subset=["Production"],inplace=True)
```

```python
df1.isnull().mean()*100
```

```python
df1.count()
```

```python
df1["District_Name"]=df1["District_Name"].str.title()
df1
```

```python
df1["Yield"] = round(df1["Production"]/df1["Area"],2)
df1
```

```python
df2=df1.drop(columns=["Production", "Area"])
df2
```

```python
df2.columns
```

```python
categorical_col = ['District_Name', 'Season', 'Crop']
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("encoding", OneHotEncoder(sparse_output=False, handle_unknown='ignore'), categorical_col),
        ("scailing", RobustScaler(), ['Crop_Year'])
    ],
    remainder='passthrough'
)
```

```python
pipeline = Pipeline(
    steps=[
        ("preproccessor",preprocessor),
        ("xgb", XGBRegressor(

            n_estimators=300,
            max_depth = 3,
            learning_rate=0.03,
            reg_alpha = 0.1,
            reg_lambda=1.5,
            subsample=0.7,
            colsample_bytree=0.7,
            n_jobs=-1,
            random_state = 42,
            min_child_weight=2,

        ))
    ]
)
```

```python
X=df2.drop(columns=["Yield"])
X
```

```python
y = df2["Yield"]
y
```

```python
X_train , X_test , y_train , y_test = train_test_split(X , y ,test_size=0.3 ,random_state=42)
```

```python
pipeline.fit(X_train,y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python

print(f"- Mean Absolute Error (MAE): {mean_absolute_error(y_test, y_pred):.3f}")
print(f"- Root Mean Squared Error (RMSE): {root_mean_squared_error(y_test, y_pred):.3f}")
print(f"- R2 Score: {r2_score(y_test, y_pred)*100:.3f}%")
```

```python
param_grid = {
    'xgb__n_estimators': [100, 200, 300 ],
    'xgb__max_depth': [3,2],
    'xgb__learning_rate': [0.01, 0.03,],
    'xgb__subsample': [0.7, 0.75, 0.6,0.8]
}

from sklearn.model_selection import GridSearchCV

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring='r2',
    cv=3,
    n_jobs=-1,
    verbose=2
)

print("Starting GridSearchCV...")
grid_search.fit(X_train, y_train)
print("GridSearchCV completed.")
```

```python
print(f"Best parameters found: {grid_search.best_params_}")
print(f"Best R2 score from cross-validation: {grid_search.best_score_:.3f}")


best_model = grid_search.best_estimator_
```

```python
y_pred_tuned = best_model.predict(X_test)
y_pred_train = best_model.predict(X_train)

print(f"\nPerformance of the Tuned Model on the Train Set:")
print(mean_absolute_error(y_train, y_pred_train))
print(root_mean_squared_error(y_train, y_pred_train))
print(r2_score(y_train, y_pred_train) * 100,"\n")

mae_tuned = mean_absolute_error(y_test, y_pred_tuned)
rmse_tuned = root_mean_squared_error(y_test, y_pred_tuned)
r2_tuned = r2_score(y_test, y_pred_tuned) * 100

print(f"\nPerformance of the Tuned Model on the Test Set:")
print(f"- Mean Absolute Error (MAE): {mae_tuned:.3f}")
print(f"- Root Mean Squared Error (RMSE): {rmse_tuned:.3f}")
print(f"- R2 Score: {r2_tuned:.3f}%")

print("\nOriginal Model Performance ")
print(f"- Mean Absolute Error (MAE): {mean_absolute_error(y_test, y_pred):.3f}")
print(f"- Root Mean Squared Error (RMSE): {root_mean_squared_error(y_test, y_pred):.3f}")
print(f"- R2 Score: {r2_score(y_test, y_pred)*100:.3f}%")
```
